## 1. Descripcion del Dataset

El dataset esta compuesto por 96 imagenes reales capturadas en campo en el Bosque El Olivar, San Isidro, Lima - Peru. Las imagenes corresponden a hojas de olivo con y sin presencia de mosca blanca (Aleurothrixus floccosus), capturadas con registro de georreferenciacion en diferentes condiciones de iluminacion.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from PIL import Image

DATA_DIR = Path('data/raw')
INFESTADAS_DIR = DATA_DIR / 'infestadas'
SANAS_DIR = DATA_DIR / 'sanas'

print('Directorios configurados correctamente.')
```

## 2. Conteo de imagenes por clase

```python
extensiones = ['.jpg', '.jpeg', '.png']
infestadas = [f for f in INFESTADAS_DIR.iterdir() if f.suffix.lower() in extensiones]
sanas = [f for f in SANAS_DIR.iterdir() if f.suffix.lower() in extensiones]
print(f'Imagenes infestadas : {len(infestadas)}')
print(f'Imagenes sanas      : {len(sanas)}')
print(f'Total               : {len(infestadas) + len(sanas)}')
```

## 3. Distribucion de clases

```python
clases = ['Infestadas', 'Sanas']
cantidades = [len(infestadas), len(sanas)]
plt.figure(figsize=(8, 5))
plt.bar(clases, cantidades, color=['#c0392b', '#27ae60'])
plt.title('Distribucion de imagenes por clase')
plt.xlabel('Clase')
plt.ylabel('Numero de imagenes')
plt.tight_layout()
plt.show()
```

## 4. Riesgos identificados

**Desbalance de clases:** El dataset tiene mas imagenes infestadas que sanas. Se aplicara aumento de datos para balancear.

**Variabilidad de iluminacion:** Las imagenes fueron capturadas en diferentes condiciones de luz natural del Bosque El Olivar.

**Tamano heterogeneo:** Las imagenes tienen diferentes resoluciones. Se estandarizaran a 640x640 pixeles en el preprocesamiento.
