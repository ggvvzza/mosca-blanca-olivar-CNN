## 1. Descripcion del Dataset
El dataset esta compuesto por imagenes reales capturadas en campo en el Bosque El Olivar, San Isidro, Lima - Peru. Las imagenes corresponden a hojas de olivo (Olea europaea) con y sin presencia de mosca blanca (Aleurothrixus floccosus), capturadas con registro de georreferenciacion en diferentes condiciones de iluminacion y posiciones del arbol.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from PIL import Image

# Directorios del dataset
DATA_DIR = Path('data/raw')
INFESTADAS_DIR = DATA_DIR / 'infestadas'
SANAS_DIR = DATA_DIR / 'sanas'

print('Directorios configurados correctamente.')
```

## 2. Conteo de imagenes por clase

```python
extensiones = ['.jpg', '.jpeg', '.png']

infestadas = [f for f in INFESTADAS_DIR.iterdir() if f.suffix.lower() in extensiones]
sanas = [f for f in SANAS_DIR.iterdir() if f.suffix.lower() in extensiones]

print(f'Imagenes infestadas : {len(infestadas)}')
print(f'Imagenes sanas      : {len(sanas)}')
print(f'Total               : {len(infestadas) + len(sanas)}')
```

## 3. Distribucion de clases

```python
clases = ['Infestadas', 'Sanas']
cantidades = [len(infestadas), len(sanas)]

plt.figure(figsize=(8, 5))
plt.bar(clases, cantidades, color=['#c0392b', '#27ae60'])
plt.title('Distribucion de imagenes por clase')
plt.xlabel('Clase')
plt.ylabel('Numero de imagenes')
plt.tight_layout()
plt.savefig('results/distribucion_clases.png')
plt.show()
print('Grafica guardada en results/distribucion_clases.png')
```

## 4. Analisis de dimensiones de imagenes

```python
anchos = []
altos = []

for img_path in infestadas + sanas:
    with Image.open(img_path) as img:
        ancho, alto = img.size
        anchos.append(ancho)
        altos.append(alto)

print(f'Ancho promedio  : {np.mean(anchos):.0f} px')
print(f'Alto promedio   : {np.mean(altos):.0f} px')
print(f'Ancho minimo    : {np.min(anchos)} px')
print(f'Ancho maximo    : {np.max(anchos)} px')
```

## 5. Riesgos identificados
- **Desbalance de clases**: El dataset inicial cuenta con mas imagenes infestadas que sanas, lo cual puede afectar el entrenamiento del modelo. Se aplicaran tecnicas de aumento de datos para balancear las clases.
- **Variabilidad de iluminacion**: Las imagenes fueron capturadas en diferentes condiciones de luz natural, lo cual puede introducir variabilidad en el entrenamiento.
- **Tamano de imagenes heterogeneo**: Las imagenes capturadas con diferentes dispositivos presentan distintas resoluciones. Se estandarizaran a 640x640 pixeles en el preprocesamiento.
